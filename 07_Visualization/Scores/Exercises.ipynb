### Step 1. Import the necessary libraries

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

### Step 2. Create the DataFrame that should look like the one below.

```python
column = ['first_name', 'last_name' ,'age', 'female', 'preTestScore', 'postTestScore']
data = [
        ['Jason', 'Miller', 42, 0, 4, 25],
        ['Molly', 'Jacobson', 52, 1, 24, 94],
        ['Tina', 'Ali', 36, 1, 31, 57],
        ['Jake', 'Milner', 24, 0, 2, 62],
        ['Amy', 'Cooze', 73, 1, 3, 70]        
       ]

df = pd.DataFrame(data, columns=column)
df
```

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

```python
plt.scatter(df.preTestScore, df.postTestScore, s=df.age)
plt.title("preTestScore vs postTestScore")
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
```

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

```python
plt.scatter(df.preTestScore, df.postTestScore, s=df.postTestScore * 4.5, c=df.female)
plt.title("preTestScore vs postTestScore")
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
```

### BONUS: Create your own question and answer it.

```python
plt.figure(figsize=(8,6))
colors = df['female'].map({1: 'pink', 0: 'blue'})  # Female: pink, Male: blue

scatter = plt.scatter(df['preTestScore'], df['postTestScore'],
                      s=df['age'] * 5,  # Scale age for better visibility
                      c=colors,
                      alpha=0.6,
                      edgecolors='w')

plt.xlabel('Pre-Test Score')
plt.ylabel('Post-Test Score')
plt.title('Pre-Test vs Post-Test Scores (Size=Age, Color=Gender)')

# Create a legend manually
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='Female', markerfacecolor='pink', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Male', markerfacecolor='blue', markersize=10)
]
plt.legend(handles=legend_elements, title='Gender')

plt.grid(True)
plt.show()
```
